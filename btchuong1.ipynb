### bài tập 1 
Kiểu dữ liệu set trong Python và FiniteSet trong gói sympy đều dùng để lưu trữ các tập hợp phần tử không trùng lặp, nhưng phục vụ hai mục đích khác nhau (lập trình tổng quát và toán học ký hiệu)
#### giống nhau:
Không chứa phần tử trùng lặp: Cả hai cấu trúc đều tự động loại bỏ các phần tử bị lặp lại.

Không quan tâm đến thứ tự: Các phần tử bên trong không sắp xếp theo một chỉ mục (index) cố định

Hỗ trợ toán tử tập hợp cơ bản: Đều cho phép thực hiện các phép toán như hợp (union), giao (intersection) và hiệu (difference

#### khác nhau
##### set (python)
Tối ưu tốc độ xử lý, lưu trữ dữ liệu và tìm kiếm trong lập trình.

Là kiểu mutable (có thể thêm, xóa phần tử bằng .add(), .remove()).

Bắt buộc phải là kiểu dữ liệu có thể băm (hashable như số, chuỗi, tuple). Không chứa được set khác.

Chỉ hỗ trợ các phép toán tập hợp cơ bản.

##### finiteset (sympy)
Dùng cho toán học ký hiệu (symbolic math), đại số và giải tích.

Là kiểu immutable (không thể thay đổi sau khi tạo, hoạt động dựa trên frozenset)

Chứa được các biểu thức toán học phức tạp, biến ký hiệu (Symbol), hoặc thậm chí chứa một FiniteSet khác bên trong.

Hỗ trợ thêm tích Descartes (*, **), tập con nguồn (powerset()), và rút gọn biểu thức.

#### bài tập 3

```python
def systolic_level(s):
    if s < 90:
        return 0
    elif s < 120:
        return 1
    elif s <= 129:
        return 2
    elif s <= 139:
        return 3
    elif s <= 159:
        return 4
    elif s <= 179:
        return 5
    else:
        return 6

def diastolic_level(d):
    if d < 60:
        return 0
    elif d < 80:
        return 1
    elif d <= 84:
        return 2
    elif d <= 89:
        return 3
    elif d <= 99:
        return 4
    elif d <= 109:
        return 5
    else:
        return 6

labels = [
    "Thap (hypotension)",
    "Toi nhat (optimal)",
    "Binh thuong (normal)",
    "Binh thuong cao",
    "THA do 1 (nhe)",
    "THA do 2 (trung binh)",
    "THA do 3 (nang)"
]

def classify(systolic, diastolic):
    if systolic >= 140 and diastolic < 90:
        return "THA tam thu don doc"
    level = max(systolic_level(systolic), diastolic_level(diastolic))
    return labels[level]

print(classify(85, 55))
print(classify(125, 105)) 
```

```python
import os
import sys

def split_file(input_path, num_disks, block_size, mode="raid5"):
    with open(input_path, "rb") as f:
        data = f.read()

    original_size = len(data)
    stripe_data_size = block_size * (num_disks - 1)
    pad = (-len(data)) % stripe_data_size
    data += b"\x00" * pad

    num_stripes = len(data) // stripe_data_size
    disks = [bytearray() for _ in range(num_disks)]

    for s in range(num_stripes):
        stripe = data[s * stripe_data_size:(s + 1) * stripe_data_size]
        blocks = [stripe[i * block_size:(i + 1) * block_size] for i in range(num_disks - 1)]

        if mode == "raid5":
            parity_disk = s % num_disks
        else:
            parity_disk = num_disks - 1

        parity = bytearray(block_size)
        bi = 0
        for d in range(num_disks):
            if d == parity_disk:
                continue
            block = blocks[bi]
            bi += 1
            disks[d].extend(block)
            for i in range(block_size):
                parity[i] ^= block[i]

        disks[parity_disk].extend(parity)

    meta = {
        "original_size": original_size,
        "num_disks": num_disks,
        "block_size": block_size,
        "num_stripes": num_stripes,
        "mode": mode,
    }
    return disks, meta

def write_disks(disks, meta, out_dir):
    os.makedirs(out_dir, exist_ok=True)
    for i, d in enumerate(disks):
        with open(os.path.join(out_dir, f"disk{i}.bin"), "wb") as f:
            f.write(bytes(d))
    with open(os.path.join(out_dir, "meta.txt"), "w") as f:
        for k, v in meta.items():
            f.write(f"{k}={v}\n")

def read_disks(out_dir, meta):
    disks = []
    for i in range(meta["num_disks"]):
        with open(os.path.join(out_dir, f"disk{i}.bin"), "rb") as f:
            disks.append(f.read())
    return disks

def read_meta(out_dir):
    meta = {}
    with open(os.path.join(out_dir, "meta.txt")) as f:
        for line in f:
            k, v = line.strip().split("=")
            meta[k] = v
    meta["original_size"] = int(meta["original_size"])
    meta["num_disks"] = int(meta["num_disks"])
    meta["block_size"] = int(meta["block_size"])
    meta["num_stripes"] = int(meta["num_stripes"])
    return meta

def reconstruct(out_dir, failed_disk, output_path):
    meta = read_meta(out_dir)
    num_disks = meta["num_disks"]
    block_size = meta["block_size"]
    num_stripes = meta["num_stripes"]
    mode = meta["mode"]

    disks = []
    for i in range(num_disks):
        if i == failed_disk:
            disks.append(None)
        else:
            with open(os.path.join(out_dir, f"disk{i}.bin"), "rb") as f:
                disks.append(f.read())

    recovered = bytearray(num_stripes * block_size)
    for s in range(num_stripes):
        xor_block = bytearray(block_size)
        for d in range(num_disks):
            if d == failed_disk:
                continue
            block = disks[d][s * block_size:(s + 1) * block_size]
            for i in range(block_size):
                xor_block[i] ^= block[i]
        recovered[s * block_size:(s + 1) * block_size] = xor_block

    with open(os.path.join(out_dir, f"disk{failed_disk}_recovered.bin"), "wb") as f:
        f.write(bytes(recovered))

    return meta

def rebuild_original(out_dir, output_path, recovered_disk=None):
    meta = read_meta(out_dir)
    num_disks = meta["num_disks"]
    block_size = meta["block_size"]
    num_stripes = meta["num_stripes"]
    mode = meta["mode"]

    disks = []
    for i in range(num_disks):
        if recovered_disk is not None and i == recovered_disk:
            with open(os.path.join(out_dir, f"disk{i}_recovered.bin"), "rb") as f:
                disks.append(f.read())
        else:
            with open(os.path.join(out_dir, f"disk{i}.bin"), "rb") as f:
                disks.append(f.read())

    result = bytearray()
    for s in range(num_stripes):
        if mode == "raid5":
            parity_disk = s % num_disks
        else:
            parity_disk = num_disks - 1
        for d in range(num_disks):
            if d == parity_disk:
                continue
            result.extend(disks[d][s * block_size:(s + 1) * block_size])

    result = result[:meta["original_size"]]
    with open(output_path, "wb") as f:
        f.write(bytes(result))

if __name__ == "__main__":
    input_file = sys.argv[1] if len(sys.argv) > 1 else "input.txt"
    out_dir = "raid_out"
    num_disks = 4
    block_size = 16
    mode = "raid5"

    disks, meta = split_file(input_file, num_disks, block_size, mode)
    write_disks(disks, meta, out_dir)
    print(f"Split into {num_disks} disks, mode={mode}, stripes={meta['num_stripes']}")

    failed_disk = 1
    reconstruct(out_dir, failed_disk, out_dir)
    print(f"Reconstructed disk {failed_disk}")

    rebuild_original(out_dir, "rebuilt.txt", recovered_disk=failed_disk)

    with open(input_file, "rb") as f1, open("rebuilt.txt", "rb") as f2:
        ok = f1.read() == f2.read()
    print("Rebuild matches original:", ok)
```

#### bài tập 8

```python
import os
import sys

GF_EXP = [0] * 512
GF_LOG = [0] * 256

def _init_gf():
    poly = 0x11d
    x = 1
    for i in range(255):
        GF_EXP[i] = x
        GF_LOG[x] = i
        x <<= 1
        if x & 0x100:
            x ^= poly
    for i in range(255, 512):
        GF_EXP[i] = GF_EXP[i - 255]

_init_gf()

def gf_mul(a, b):
    if a == 0 or b == 0:
        return 0
    return GF_EXP[GF_LOG[a] + GF_LOG[b]]

def gf_pow(a, power):
    if a == 0:
        return 0
    return GF_EXP[(GF_LOG[a] * power) % 255]

def gf_inv(a):
    return GF_EXP[255 - GF_LOG[a]]

def split_file(input_path, num_data_disks, block_size):
    with open(input_path, "rb") as f:
        data = f.read()

    original_size = len(data)
    stripe_data_size = block_size * num_data_disks
    pad = (-len(data)) % stripe_data_size
    data += b"\x00" * pad
    num_stripes = len(data) // stripe_data_size

    data_disks = [bytearray() for _ in range(num_data_disks)]
    p_disk = bytearray()
    q_disk = bytearray()

    for s in range(num_stripes):
        stripe = data[s * stripe_data_size:(s + 1) * stripe_data_size]
        blocks = [stripe[i * block_size:(i + 1) * block_size] for i in range(num_data_disks)]

        p = bytearray(block_size)
        q = bytearray(block_size)
        for d in range(num_data_disks):
            data_disks[d].extend(blocks[d])
            coeff = gf_pow(2, d)
            for i in range(block_size):
                p[i] ^= blocks[d][i]
                q[i] ^= gf_mul(coeff, blocks[d][i])

        p_disk.extend(p)
        q_disk.extend(q)

    meta = {
        "original_size": original_size,
        "num_data_disks": num_data_disks,
        "block_size": block_size,
        "num_stripes": num_stripes,
    }
    return data_disks, p_disk, q_disk, meta

def write_disks(data_disks, p_disk, q_disk, meta, out_dir):
    os.makedirs(out_dir, exist_ok=True)
    for i, d in enumerate(data_disks):
        with open(os.path.join(out_dir, f"disk{i}.bin"), "wb") as f:
            f.write(bytes(d))
    with open(os.path.join(out_dir, "diskP.bin"), "wb") as f:
        f.write(bytes(p_disk))
    with open(os.path.join(out_dir, "diskQ.bin"), "wb") as f:
        f.write(bytes(q_disk))
    with open(os.path.join(out_dir, "meta.txt"), "w") as f:
        for k, v in meta.items():
            f.write(f"{k}={v}\n")

def read_meta(out_dir):
    meta = {}
    with open(os.path.join(out_dir, "meta.txt")) as f:
        for line in f:
            k, v = line.strip().split("=")
            meta[k] = v
    meta["original_size"] = int(meta["original_size"])
    meta["num_data_disks"] = int(meta["num_data_disks"])
    meta["block_size"] = int(meta["block_size"])
    meta["num_stripes"] = int(meta["num_stripes"])
    return meta

def read_disk(out_dir, name):
    with open(os.path.join(out_dir, f"{name}.bin"), "rb") as f:
        return f.read()

def reconstruct_two_failed(out_dir, failed_a, failed_b):
    meta = read_meta(out_dir)
    n = meta["num_data_disks"]
    bs = meta["block_size"]
    stripes = meta["num_stripes"]

    names = [f"disk{i}" for i in range(n)] + ["diskP", "diskQ"]
    disks = {}
    for name in names:
        if name in (failed_a, failed_b):
            disks[name] = None
        else:
            disks[name] = read_disk(out_dir, name)

    recovered = {failed_a: bytearray(stripes * bs), failed_b: bytearray(stripes * bs)}

    for s in range(stripes):
        p = bytearray(disks["diskP"][s * bs:(s + 1) * bs]) if disks["diskP"] is not None else None
        q = bytearray(disks["diskQ"][s * bs:(s + 1) * bs]) if disks["diskQ"] is not None else None

        data_indices = list(range(n))
        known = {}
        for i in data_indices:
            name = f"disk{i}"
            if disks[name] is not None:
                known[i] = disks[name][s * bs:(s + 1) * bs]

        if "diskP" not in (failed_a, failed_b) and "diskQ" not in (failed_a, failed_b):
            xa = int(failed_a.replace("disk", ""))
            xb = int(failed_b.replace("disk", ""))
            ga, gb = gf_pow(2, xa), gf_pow(2, xb)
            p_known = bytearray(bs)
            q_known = bytearray(bs)
            for i, blk in known.items():
                coeff = gf_pow(2, i)
                for j in range(bs):
                    p_known[j] ^= blk[j]
                    q_known[j] ^= gf_mul(coeff, blk[j])

            for j in range(bs):
                p_rem = p[j] ^ p_known[j]
                q_rem = q[j] ^ q_known[j]
                denom = ga ^ gb
                da = gf_mul(q_rem ^ gf_mul(gb, p_rem), gf_inv(denom))
                db = p_rem ^ da
                recovered[failed_a][s * bs + j] = da
                recovered[failed_b][s * bs + j] = db

        elif "diskQ" in (failed_a, failed_b):
            missing_data = failed_a if failed_a != "diskQ" else failed_b
            xi = int(missing_data.replace("disk", ""))
            xor = bytearray(bs)
            for i, blk in known.items():
                for j in range(bs):
                    xor[j] ^= blk[j]
            for j in range(bs):
                recovered[missing_data][s * bs + j] = p[j] ^ xor[j]

            qcalc = bytearray(bs)
            for i, blk in known.items():
                coeff = gf_pow(2, i)
                for j in range(bs):
                    qcalc[j] ^= gf_mul(coeff, blk[j])
            coeff_missing = gf_pow(2, xi)
            for j in range(bs):
                term = gf_mul(coeff_missing, recovered[missing_data][s * bs + j])
                recovered["diskQ"][s * bs + j] = qcalc[j] ^ term

        elif "diskP" in (failed_a, failed_b):
            missing_data = failed_a if failed_a != "diskP" else failed_b
            xi = int(missing_data.replace("disk", ""))
            qcalc = bytearray(bs)
            for i, blk in known.items():
                coeff = gf_pow(2, i)
                for j in range(bs):
                    qcalc[j] ^= gf_mul(coeff, blk[j])
            inv_coeff = gf_inv(gf_pow(2, xi))
            for j in range(bs):
                diff = q[j] ^ qcalc[j]
                recovered[missing_data][s * bs + j] = gf_mul(diff, inv_coeff)

            xor = bytearray(bs)
            for i, blk in known.items():
                for j in range(bs):
                    xor[j] ^= blk[j]
            for j in range(bs):
                recovered["diskP"][s * bs + j] = xor[j] ^ recovered[missing_data][s * bs + j]

    for name, content in recovered.items():
        with open(os.path.join(out_dir, f"{name}_recovered.bin"), "wb") as f:
            f.write(bytes(content))

    return meta

def rebuild_original(out_dir, output_path, recovered_names=()):
    meta = read_meta(out_dir)
    n = meta["num_data_disks"]
    bs = meta["block_size"]

    blocks = []
    for i in range(n):
        name = f"disk{i}"
        if name in recovered_names:
            blocks.append(read_disk(out_dir, f"{name}_recovered"))
        else:
            blocks.append(read_disk(out_dir, name))

    result = bytearray()
    stripes = meta["num_stripes"]
    for s in range(stripes):
        for d in range(n):
            result.extend(blocks[d][s * bs:(s + 1) * bs])

    result = result[:meta["original_size"]]
    with open(output_path, "wb") as f:
        f.write(bytes(result))

if __name__ == "__main__":
    input_file = sys.argv[1] if len(sys.argv) > 1 else "input.txt"
    out_dir = "raid6_out"
    num_data_disks = 4
    block_size = 16

    data_disks, p_disk, q_disk, meta = split_file(input_file, num_data_disks, block_size)
    write_disks(data_disks, p_disk, q_disk, meta, out_dir)
    print(f"Split into {num_data_disks} data disks + P + Q, stripes={meta['num_stripes']}")

    failed_a, failed_b = "disk1", "disk2"
    reconstruct_two_failed(out_dir, failed_a, failed_b)
    print(f"Reconstructed {failed_a} and {failed_b}")

    rebuild_original(out_dir, "rebuilt.txt", recovered_names={failed_a, failed_b})

    with open(input_file, "rb") as f1, open("rebuilt.txt", "rb") as f2:
        ok = f1.read() == f2.read()
    print("Rebuild matches original:", ok)
```
